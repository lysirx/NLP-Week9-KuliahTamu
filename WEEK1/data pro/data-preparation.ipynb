````python
import pandas as pd
from google_play_scraper import Sort, reviews

result, continuation_token = reviews(
    'com.mrt.jakarta',
    lang='id',  # Bahasa Indonesia
    country='id',  # Sumber Data Review Google Play Store Indonesia
    sort=Sort.RATING,  # Rating tertinggi/terendah
    count=100,  # Ambil 100 ulasan
)

# Jika ingin melanjutkan pengambilan data
if continuation_token:
    additional_result, _ = reviews(
        'com.mrt.jakarta',
        continuation_token=continuation_token
    )
    result.extend(additional_result)

# Konversi ke DataFrame
df = pd.DataFrame(result)

# # Simpan ke CSV
df.to_csv('DatasetMentahMRT.csv', index=False, encoding='utf-8')
````

### 2.2.0 Case Folding
```python
def case_folding(teks):
    return teks.lower()
```

Case folding adalah proses mengubah semua huruf menjadi **huruf kecil (lowercase)** supaya tidak ada perbedaan antara `"Bagus"`, `"BAGUS"`, dan `"bagus"` — ketiganya dianggap kata yang sama.

Contoh:
```
Input  : "Aplikasi MRT Sangat BAGUS"
Output : "aplikasi mrt sangat bagus"
```

> Ini langkah pertama yang wajib dilakukan sebelum proses cleaning lainnya, karena pada saat cleaning data fungsi regex dan pencocokan kata akan lebih konsisten kalau semua huruf sudah kecil semua.

### 2.2.1 Hapus Special Karakter
# 📋 Regex Cheatsheet — Python `re` Library
> Panduan lengkap pola regex yang sering dipakai untuk text cleaning 
---

## 7. Cheatsheet Ringkas — Untuk NLP / Text Cleaning

| Kebutuhan | Regex | Fungsi |
|---|---|---|
| Hapus angka | `\d+` | `re.sub` |
| Hapus huruf | `[a-zA-Z]+` | `re.sub` |
| Hapus tanda baca | `[^\w\s]` | `re.sub` |
| Hapus URL | `http\S+\|www\.\S+` | `re.sub` |
| Hapus tag HTML | `<.*?>` | `re.sub` |
| Hapus tab & newline | `[\t\n\\]` | `re.sub` |
| Hapus spasi berlebih | `\s+` | `re.sub` + `.strip()` |
| Hapus satu karakter | `\b[a-zA-Z]\b` | `re.sub` |
| Hapus kata pendek (≤3) | `\b[a-zA-Z]{1,3}\b` | `re.sub` |
| Hapus mention (@user) | `@\w+` | `re.sub` |
| Hapus hashtag (#tag) | `#\w+` | `re.sub` |
| Hapus emoji | `[^\x00-\x7F]+` | `re.sub` |
| Cari semua angka | `\d+` | `re.findall` |
| Pecah per kata | `\s+` | `re.split` |

---

> 💡 **Tips:** Selalu gunakan raw string `r'...'` saat menulis pola regex di Python supaya backslash tidak salah diinterpretasikan.

````python
import string
import re
import nltk

nltk.download('punkt')
nltk.download('popular')
````

````python

nltk.download('popular')

# ── 1. Load Data ──────────────────────────────────────────────
df = pd.read_csv('MengolahDataMRT1.csv')
MengolahData = df[['content']].copy()

# ── 2. Fungsi Preprocessing ───────────────────────────────────

# .replace('\\t', " ")
#          ↑       ↑
#          1       2
# argumen 1 = teks/karakter yang mau dicari
# argumen 2 = teks/karakter penggantinya
def remove_review_special(text):
    """Hapus karakter escape, mention, hashtag, dan URL"""
    text = (text.replace('\\t', " ")
                .replace('\\n', " ")
                .replace('\\u', " ")
                .replace('\\',  ""))
    text = text.encode('ascii', 'replace').decode('ascii')
    text = ' '.join(re.sub(r"([@#][A-Za-z0-9]+)|(\w+:\/\/\S+)", " ", text).split())
    text = text.replace("http://", " ").replace("https://", " ")
    return text

# re.sub(pattern, replacement, text)
#          ↑         ↑         ↑
#          1         2         3
# argumen 1 = pattern/pola yang mau dicari
# argumen 2 = penggantinya, di sini diganti dengan spasi
# argumen 3 = teks yang mau diproses
def remove_number(text):
    """Hapus semua angka"""
    return re.sub(r"\d+", "", text)

# str.maketrans("", "", string.punctuation)
#               ↑   ↑          ↑
#               1   2          3
# argumen 1 "" = karakter yang mau diganti
# argumen 2 "" = karakter penggantinya
# argumen 3 string.punctuation = karakter yang mau dihapus
def remove_punctuation(text):
    """Hapus semua tanda baca"""
    return text.translate(str.maketrans("", "", string.punctuation))

def remove_whitespace_LT(text):
    """Hapus spasi di awal dan akhir teks"""
    return text.strip()

def remove_whitespace_multiple(text):
    """Hapus spasi berlebih di tengah teks"""
    return re.sub(r'\s+', ' ', text)

def remove_single_char(text):
    """Hapus kata yang hanya terdiri dari 1 karakter"""
    return re.sub(r"\b[a-zA-Z]\b", "", text)

def remove_small_words(text):
    """Hapus kata pendek (1-3 karakter) kecuali kata yang dikecualikan"""
    exceptions = ["kai", "tim"]
    pattern = r'\b(?!' + '|'.join(exceptions) + r')\w{1,3}\b'
    return re.sub(pattern, '', text)

# ── 3. Terapkan Semua Fungsi ──────────────────────────────────

cleaning_pipeline = [
    remove_review_special,
    remove_number,
    remove_punctuation,
    remove_whitespace_LT,
    remove_whitespace_multiple,
    remove_single_char,
    remove_small_words,
]

MengolahData['remove punctuation'] = MengolahData['content']
for func in cleaning_pipeline:
    MengolahData['remove punctuation'] = MengolahData['remove punctuation'].apply(func)

# ── 4. Case Folding ───────────────────────────────────────────

MengolahData['case folding']        = MengolahData['content'].str.lower()
MengolahData['remove punctuation']  = MengolahData['remove punctuation'].str.lower()

# ── 5. Susun Kolom & Simpan ───────────────────────────────────

final_data = MengolahData[['content', 'case folding', 'remove punctuation']]
final_data.to_csv('Data-Punctualization-MRT.csv', index=False)

print("Data berhasil disimpan ke Data-Punctualization-MRT.csv")
print(f"   Total baris: {len(final_data)}")
print()
print(final_data.head())
````
