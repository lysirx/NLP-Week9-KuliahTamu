### 2.2.1 Hapus Special Karakter
# 📋 Regex Cheatsheet — Python `re` Library
> Panduan lengkap pola regex yang sering dipakai untuk text cleaning 
---

## 📌 Fungsi Utama `re`

| Fungsi | Kegunaan | Contoh |
|---|---|---|
| `re.sub(pola, ganti, teks)` | Cari & ganti semua pola | `re.sub(r'\d+', '', teks)` |
| `re.findall(pola, teks)` | Cari semua yang cocok, return list | `re.findall(r'\d+', teks)` |
| `re.search(pola, teks)` | Cek apakah pola ada di teks | `re.search(r'halo', teks)` |
| `re.match(pola, teks)` | Cek pola hanya di awal teks | `re.match(r'^halo', teks)` |
| `re.split(pola, teks)` | Pecah teks berdasarkan pola | `re.split(r'\s+', teks)` |
| `re.compile(pola)` | Simpan pola supaya bisa dipakai ulang | `p = re.compile(r'\d+')` |

---

## 1. Karakter Dasar

| Pola | Arti | Contoh Input | Yang Cocok |
|---|---|---|---|
| `\d` | satu angka (0–9) | `"abc123"` | `1`, `2`, `3` |
| `\d+` | satu atau lebih angka | `"abc123"` | `123` |
| `\D` | bukan angka | `"abc123"` | `a`, `b`, `c` |
| `\w` | huruf, angka, underscore | `"hi_2!"` | `h`, `i`, `_`, `2` |
| `\w+` | satu atau lebih huruf/angka/underscore | `"hi_2!"` | `hi_2` |
| `\W` | bukan huruf/angka/underscore | `"hi!"` | `!` |
| `\s` | spasi, tab, newline | `"a b"` | spasi |
| `\s+` | satu atau lebih whitespace | `"a   b"` | `   ` |
| `\S` | bukan spasi | `"a b"` | `a`, `b` |
| `.` | karakter apa saja kecuali newline | `"ab"` | `a`, `b` |
| `.*` | semua karakter (nol atau lebih) | `"halo dunia"` | `halo dunia` |
| `.+` | semua karakter (satu atau lebih) | `"halo"` | `halo` |

---

## 2. Kuantifier (Jumlah Pengulangan)

| Pola | Arti | Contoh |
|---|---|---|
| `+` | satu atau lebih | `\d+` → cocok `1`, `12`, `123` |
| `*` | nol atau lebih | `\d*` → cocok `""` atau `123` |
| `?` | nol atau satu (opsional) | `colou?r` → cocok `color` atau `colour` |
| `{n}` | tepat n kali | `\d{3}` → cocok `123` |
| `{n,}` | minimal n kali | `\d{2,}` → cocok `12`, `123`, `1234` |
| `{n,m}` | antara n sampai m kali | `\d{1,3}` → cocok `1`, `12`, `123` |
| `+?` | satu atau lebih (non-greedy) | `\d+?` → ambil sesedikit mungkin |
| `*?` | nol atau lebih (non-greedy) | `.*?` → ambil sesedikit mungkin |

---

## 3. Posisi Teks (Anchors)

| Pola | Arti | Contoh |
|---|---|---|
| `^` | awal teks / awal baris | `^halo` → cocok kalau teks dimulai `halo` |
| `$` | akhir teks / akhir baris | `bagus$` → cocok kalau teks diakhiri `bagus` |
| `\b` | batas kata (word boundary) | `\bkata\b` → cocok `kata` tapi bukan `katakata` |
| `\B` | bukan batas kata | `\Bkat\B` → cocok `kat` di tengah kata |
| `\A` | benar-benar awal string | `\Ahalo` → hanya di awal string |
| `\Z` | benar-benar akhir string | `bagus\Z` → hanya di akhir string |

---

## 4. Kelompok & Pilihan

| Pola | Arti | Contoh |
|---|---|---|
| `[abc]` | salah satu dari a, b, atau c | `[aeiou]` → cocok huruf vokal |
| `[^abc]` | bukan a, b, atau c | `[^aeiou]` → cocok konsonan |
| `[a-z]` | huruf kecil a sampai z | `[a-z]+` → cocok kata huruf kecil |
| `[A-Z]` | huruf besar A sampai Z | `[A-Z]+` → cocok kata huruf besar |
| `[a-zA-Z]` | semua huruf | `[a-zA-Z]+` → cocok semua huruf |
| `[0-9]` | angka 0 sampai 9 (sama dengan `\d`) | `[0-9]+` → cocok angka |
| `[^\w\s]` | bukan huruf dan bukan spasi | cocok semua tanda baca & simbol |
| `a\|b` | a atau b | `kucing\|anjing` → cocok salah satu |
| `(abc)` | kelompokkan pola | `(ha)+` → cocok `ha`, `haha`, `hahaha` |
| `(?:abc)` | kelompok tanpa capture | `(?:ha)+` → sama tapi tidak disimpan |

---

## 5. Karakter Khusus (Escape)

| Pola | Arti |
|---|---|
| `\t` | tab |
| `\n` | newline / enter |
| `\r` | carriage return |
| `\\` | backslash literal |
| `\.` | titik literal (bukan wildcard) |
| `\*` | bintang literal |
| `\+` | plus literal |
| `\?` | tanda tanya literal |
| `\(` `\)` | kurung literal |
| `\[` `\]` | kurung kotak literal |
| `\{` `\}` | kurung kurawal literal |

---

## 6. Contoh Penggunaan untuk Text Cleaning

```python
import re

teks = "Halo! ini review MRT 😊 http://mrt.co.id harga 5000\n\t keren!!!"

# Hapus angka
re.sub(r'\d+', '', teks)
# → "Halo! ini review MRT 😊 http://mrt.co.id harga \n\t keren!!!"

# Hapus URL
re.sub(r'http\S+|www\.\S+', '', teks)
# → "Halo! ini review MRT 😊  harga 5000\n\t keren!!!"

# Hapus tanda baca dan simbol
re.sub(r'[^\w\s]', '', teks)
# → "Halo ini review MRT  httpmrtcoid harga 5000 keren"

# Hapus tab dan newline
re.sub(r'[\t\n]', ' ', teks)
# → "Halo! ini review MRT 😊 http://mrt.co.id harga 5000   keren!!!"

# Hapus spasi berlebih
re.sub(r'\s+', ' ', teks).strip()
# → "Halo! ini review MRT 😊 http://mrt.co.id harga 5000 keren!!!"

# Hapus satu karakter
re.sub(r'\b[a-zA-Z]\b', '', teks)
# → "Halo! ini review MRT 😊 http://mrt.co.id harga 5000\n\t keren!!!"

# Hapus kata 1–3 huruf
re.sub(r'\b[a-zA-Z]{1,3}\b', '', teks)
# → "Halo! review MRT 😊 http://mrt.co.id harga 5000\n\t keren!!!"

# Hapus semua huruf
re.sub(r'[a-zA-Z]', '', teks)
# → "! 😊 ://. 5000\n\t !!!"

# Hapus semua huruf + angka + underscore
re.sub(r'\w', '', teks)
# → "! 😊 ://. \n\t !!!"
```

## 7. Cheatsheet Ringkas — Untuk NLP / Text Cleaning

| Kebutuhan | Regex | Fungsi |
|---|---|---|
| Hapus angka | `\d+` | `re.sub` |
| Hapus huruf | `[a-zA-Z]+` | `re.sub` |
| Hapus tanda baca | `[^\w\s]` | `re.sub` |
| Hapus URL | `http\S+\|www\.\S+` | `re.sub` |
| Hapus tag HTML | `<.*?>` | `re.sub` |
| Hapus tab & newline | `[\t\n\\]` | `re.sub` |
| Hapus spasi berlebih | `\s+` | `re.sub` + `.strip()` |
| Hapus satu karakter | `\b[a-zA-Z]\b` | `re.sub` |
| Hapus kata pendek (≤3) | `\b[a-zA-Z]{1,3}\b` | `re.sub` |
| Hapus mention (@user) | `@\w+` | `re.sub` |
| Hapus hashtag (#tag) | `#\w+` | `re.sub` |
| Hapus emoji | `[^\x00-\x7F]+` | `re.sub` |

| Pecah per kata | `\s+` | `re.split` |

## Bedah Sintaks

```
re.sub( r'\d+' ,   ''   ,  teks )
           ①         ②       ③
```

Mari kita bedah satu per satu:

---

### ① `r'\d+'` — Pola Regex (Pattern)

Bagian ini terdiri dari dua hal:

---

#### Huruf `r` di depan kutip

```python
r'\d+'
```

`r` artinya **raw string** — memberitahu Python untuk **tidak menginterpretasikan backslash** `\` sebagai karakter khusus.

Contoh bedanya:

```python
'\d+'   # Python baca: karakter aneh + d+   ❌ salah
r'\d+'  # Python baca: \d+ apa adanya       ✅ benar
```

> Jadi huruf `r` wajib dipakai di depan pola regex supaya `\d`, `\s`, `\w`, dan sejenisnya tidak salah dibaca oleh Python.

---

#### `\d+` di dalam kutip

| Bagian | Arti |
|---|---|
| `\d` | cocokkan **satu angka** (0–9) |
| `+` | ulangi **satu atau lebih** kali |
| Gabungan `\d+` | cocokkan **satu atau lebih angka berurutan** |

Contoh cara kerjanya:

```python
teks = "harga 5000 rupiah tahun 2024"

# \d+ akan menemukan: "5000" dan "2024"
# lalu keduanya dihapus
```

---

### ② `''` — Pengganti (Replacement)

```python
re.sub(r'\d+', '', teks)
#              ^^
#          string kosong
```

Ini adalah **teks pengganti** — semua yang cocok dengan pola akan diganti dengan isi bagian ini.

Karena isinya kosong `''`, semua angka yang ditemukan akan **dihapus**.

Perbandingan:

```python
teks = "harga 5000 rupiah"

re.sub(r'\d+', '', teks)    # → "harga  rupiah"   ← angka dihapus
re.sub(r'\d+', 'X', teks)   # → "harga X rupiah"  ← angka diganti X
re.sub(r'\d+', '???', teks) # → "harga ??? rupiah" ← angka diganti ???
```

---

### ③ `teks` — Variabel (Tanpa Kutip)

```python
re.sub(r'\d+', '', teks)
#                   ^^^^
#               tidak pakai kutip
```

`teks` adalah **variabel** yang sudah didefinisikan sebelumnya, bukan string baru. Makanya tidak pakai kutip.

Bedanya:

```python
# Pakai variabel → tidak dikutip
teks = "harga 5000 rupiah"
re.sub(r'\d+', '', teks)

# Pakai string langsung → dikutip
re.sub(r'\d+', '', "harga 5000 rupiah")
```

Keduanya menghasilkan output yang sama. Tapi kalau datanya dari DataFrame seperti kasus kita, selalu pakai variabel.

---

## Kesimpulan

```python
re.sub( r'\d+' ,   ''   ,  teks )
#          ①         ②       ③
#       pola yg    diganti   teks
#       dicari      kosong   yang
#     (cari angka)  = hapus  diproses
```

| Bagian | Fungsi |
|---|---|
| `r` | raw string, biar `\` tidak salah dibaca Python |
| `'\d+'` | pola regex yang dicari di dalam teks |
| `''` | pengganti — kosong berarti hapus |
| `teks` | variabel yang berisi teks yang mau diproses |

> Cara bacanya: **"di dalam variabel `teks`, cari semua angka (`\d+`), lalu hapus (ganti dengan `''`)"**
